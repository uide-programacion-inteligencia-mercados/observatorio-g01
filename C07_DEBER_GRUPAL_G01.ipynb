```python
#@title PASO 2 · Correr el robot nuevo { display-mode: "form" }
#@markdown Pulsa el botón redondo (no escribes nada)
import json, os, subprocess, sys
from pathlib import Path
import pandas as pd
from IPython.display import display

ROBOT_V2 = '# ============================================================================\n# ARCHIVO: robot_banco_mundial.py   (el robot del equipo __GXX__)   · VERSIÓN 2 · SEMANA 7\n# QUÉ HACE:  Trae los dos indicadores del equipo desde el Banco Mundial, los revisa\n#            con las reglas de calidad del equipo y SOLO si pasan los guarda.\n# NOVEDADES DE LA SEMANA 7 («Avisa si falla»):\n#   1. Reintentos con espera creciente (2 s, 4 s, 8 s...).\n#   2. Control de calidad: si los datos no son creíbles, se detiene y no guarda.\n#   3. Bitácora: una línea por evento en logs/bitacora___GXX__.jsonl\n#   4. Salud: una fila por corrida en logs/salud___GXX__.csv (verde, amarillo o rojo).\n#   5. Escritura atómica: el archivo de datos nunca queda cortado a medias.\n#   6. Simulacro: con SIMULAR_FALLA=true finge que el Banco Mundial no contesta.\n# ESTO PUEDEN CAMBIARLO: las reglas de la sección «REGLAS DEL EQUIPO».\n# ============================================================================\nimport json\nimport os\nimport sys\nimport time\nfrom datetime import datetime, timezone\nfrom pathlib import Path\n\nimport pandas as pd\nimport requests\n\nEQUIPO = "__GXX__"\nINDICADORES = __INDICADORES__\n\n# --- REGLAS DEL EQUIPO (decididas en el cuaderno de la semana 7) ---\nRANGOS = __RANGOS__              # valor mínimo y máximo CREÍBLE de cada indicador\nMIN_FILAS = __MIN_FILAS__                  # menos filas que esto = llegó incompleto\nMAX_ANIOS_SIN_DATO = __MAX_ANIOS__         # si el último año tiene más antigüedad, el dato está viejo\nREINTENTOS = __REINTENTOS__                # cuántas veces más intenta si la fuente no contesta\nESPERA_INICIAL = __ESPERA__                # segundos de la primera espera; luego se duplica\n\nDESDE, HASTA = 2015, 2025\nSIMULAR_FALLA = os.environ.get("SIMULAR_FALLA", "false").lower() == "true"\nCORRIDA = os.environ.get("GITHUB_RUN_ID") or datetime.now(timezone.utc).strftime("local-%Y%m%dT%H%M%S")\nEN_GITHUB = os.environ.get("GITHUB_ACTIONS") == "true"\n\nDATA, LOGS = Path("data"), Path("logs")\nARCHIVO_DATOS = DATA / f"indicadores_{EQUIPO}.csv"\nBITACORA = LOGS / f"bitacora_{EQUIPO}.jsonl"\nSALUD = LOGS / f"salud_{EQUIPO}.csv"\nREINTENTOS_USADOS = []\n\n\nclass ErrorDeDatos(Exception):\n    """Los datos llegaron, pero no son creíbles. Reintentar no lo arregla."""\n\n\ndef anotar(nivel, evento, **detalles):\n    # Bitácora: una línea JSON por evento, siempre con los mismos campos.\n    LOGS.mkdir(exist_ok=True)\n    fila = {"momento_utc": datetime.now(timezone.utc).isoformat(timespec="seconds"), "corrida": CORRIDA,\n            "equipo": EQUIPO, "nivel": nivel, "evento": evento, **detalles}\n    with open(BITACORA, "a", encoding="utf-8") as f:\n        f.write(json.dumps(fila, ensure_ascii=False, default=str) + "\\n")\n    print(f"[{nivel:<7}] {evento}" + (" · " + ", ".join(f"{k}={v}" for k, v in detalles.items()) if detalles else ""))\n    if EN_GITHUB and nivel in ("WARNING", "ERROR"):\n        print(f"::{\'warning\' if nivel == \'WARNING\' else \'error\'} title=Robot {EQUIPO}::{evento}")\n\n\ndef traer(codigo):\n    # Pide UN indicador. Si la fuente no contesta, espera cada vez más y reintenta.\n    url = (f"https://api.worldbank.org/v2/country/ECU/indicator/{codigo}"\n           f"?format=json&date={DESDE}:{HASTA}&per_page=100")\n    for intento in range(1, REINTENTOS + 2):\n        try:\n            if SIMULAR_FALLA:\n                raise ConnectionError("SIMULACRO: el Banco Mundial no contesta (falla provocada a propósito)")\n            r = requests.get(url, timeout=30)\n            r.raise_for_status()\n            datos = r.json()\n            if len(datos) < 2 or not datos[1]:\n                raise ErrorDeDatos(f"{codigo}: el Banco Mundial respondió, pero sin datos")\n            return datos[1]\n        except ErrorDeDatos:\n            raise                                    # datos malos: no se reintenta\n        except Exception as e:\n            if intento > REINTENTOS:\n                raise ConnectionError(f"{codigo}: no contestó después de {intento} intentos ({e})")\n            pausa = ESPERA_INICIAL * 2 ** (intento - 1)\n            REINTENTOS_USADOS.append(codigo)\n            anotar("WARNING", f"{codigo}: intento {intento} falló; espero {pausa} s y reintento", error=str(e)[:150])\n            time.sleep(pausa)\n\n\ndef revisar(tabla):\n    # El control de calidad del equipo. Devuelve la lista de problemas (vacía = todo bien).\n    problemas = []\n    if tabla.empty:\n        return ["no llegó ninguna fila"]\n    if len(tabla) < MIN_FILAS:\n        problemas.append(f"solo {len(tabla)} filas (mínimo {MIN_FILAS})")\n    for codigo, nombre in INDICADORES.items():\n        parte = tabla[tabla["codigo"] == codigo]\n        if parte.empty:\n            problemas.append(f"falta el indicador {nombre}")\n            continue\n        lo, hi = RANGOS[codigo]\n        raros = parte[(parte["valor"] < lo) | (parte["valor"] > hi)]\n        if len(raros):\n            problemas.append(f"{nombre}: {len(raros)} valor(es) fuera de lo creíble ({lo} a {hi})")\n        ultimo = int(parte["anio"].max())\n        if ultimo < datetime.now(timezone.utc).year - MAX_ANIOS_SIN_DATO:\n            problemas.append(f"{nombre}: el último dato es de {ultimo} (dato viejo)")\n    return problemas\n\n\ndef guardar_atomico(tabla):\n    # Escribe un borrador y lo cambia por el archivo de verdad en un solo movimiento.\n    DATA.mkdir(exist_ok=True)\n    borrador = ARCHIVO_DATOS.with_suffix(".tmp")\n    tabla.to_csv(borrador, index=False)\n    borrador.replace(ARCHIVO_DATOS)\n\n\ndef registrar_salud(estado, motivo, filas, segundos):\n    # Una fila por corrida. Si GitHub repite la MISMA corrida, se reemplaza (idempotencia).\n    LOGS.mkdir(exist_ok=True)\n    fila = {"fecha_utc": datetime.now(timezone.utc).isoformat(timespec="seconds"), "corrida": CORRIDA,\n            "estado": estado, "motivo": motivo, "filas": filas, "segundos": round(segundos, 1)}\n    salud = pd.read_csv(SALUD) if SALUD.exists() else pd.DataFrame(columns=list(fila))\n    salud = salud[salud["corrida"].astype(str) != str(CORRIDA)]\n    pd.concat([salud, pd.DataFrame([fila])], ignore_index=True).to_csv(SALUD, index=False)\n    return salud\n\n\ndef resumen_en_github():\n    # Escribe el tablero de salud en la pantalla de la corrida de GitHub.\n    destino = os.environ.get("GITHUB_STEP_SUMMARY")\n    if not destino or not SALUD.exists():\n        return\n    s = pd.read_csv(SALUD).tail(7).iloc[::-1]\n    icono = {"verde": "🟢", "amarillo": "🟡", "rojo": "🔴"}\n    filas = "\\n".join(f"| {r.fecha_utc[:16].replace(\'T\', \' \')} | {icono.get(r.estado, \'\')} {r.estado} | {r.motivo} |"\n                      for r in s.itertuples())\n    with open(destino, "a", encoding="utf-8") as f:\n        f.write(f"## Salud del robot {EQUIPO}\\n\\n| Fecha (UTC) | Estado | Motivo |\\n|---|---|---|\\n{filas}\\n")\n\n\ndef main():\n    inicio = time.time()\n    anotar("INFO", "inicio de la corrida", simulacro=SIMULAR_FALLA)\n    filas_guardadas = 0\n    try:\n        filas = []\n        for codigo, nombre in INDICADORES.items():\n            for d in traer(codigo):\n                if d.get("value") is not None:\n                    filas.append({"anio": int(d["date"]), "codigo": codigo, "indicador": nombre, "valor": d["value"]})\n            anotar("INFO", f"{codigo}: recibido")\n        tabla = pd.DataFrame(filas, columns=["anio", "codigo", "indicador", "valor"]).sort_values(["codigo", "anio"])\n        problemas = revisar(tabla)\n        if problemas:\n            raise ErrorDeDatos("; ".join(problemas))\n        guardar_atomico(tabla)\n        filas_guardadas = len(tabla)\n        anotar("INFO", "datos aprobados y guardados", filas=filas_guardadas, archivo=str(ARCHIVO_DATOS))\n        estado, motivo = ("amarillo", "funcionó, pero tuvo que reintentar") if REINTENTOS_USADOS else ("verde", "todo bien")\n    except ErrorDeDatos as e:\n        estado, motivo = "rojo", f"datos no confiables: {e}"\n        anotar("ERROR", motivo)\n    except Exception as e:\n        estado, motivo = "rojo", f"la fuente falló: {e}"\n        anotar("ERROR", motivo)\n    if estado == "rojo" and ARCHIVO_DATOS.exists():\n        anotar("INFO", "plan B: se conservan intactos los datos de la última corrida buena")\n    registrar_salud(estado, motivo, filas_guardadas, time.time() - inicio)\n    resumen_en_github()\n    print(f"RESULTADO: {estado.upper()} · {motivo}")\n    sys.exit(1 if estado == "rojo" else 0)\n\n\nif __name__ == "__main__":\n    main()\n'
REGLAS = {"rangos": {"FP.CPI.TOTL.ZG": [-5.0, 30.0], "NE.CON.PRVT.PC.KD.ZG": [-20.0, 20.0]}, "min_filas": 10, "max_anios": 4,
          "reintentos": 3, "espera": 2}

def escribir_robot(reglas):
    texto = (ROBOT_V2.replace("__GXX__", EQUIPO).replace("__INDICADORES__", json.dumps(INDICADORES, ensure_ascii=False))
             .replace("__RANGOS__", json.dumps({k: tuple(v) for k, v in reglas["rangos"].items()}).replace("[", "(").replace("]", ")"))
             .replace("__MIN_FILAS__", str(reglas["min_filas"])).replace("__MAX_ANIOS__", str(reglas["max_anios"]))
             .replace("__REINTENTOS__", str(reglas["reintentos"])).replace("__ESPERA__", str(reglas["espera"])))
    Path("robot_banco_mundial.py").write_text(texto, encoding="utf-8")
    return texto

def correr_robot(**variables):
    env = dict(os.environ, **variables)
    r = subprocess.run([sys.executable, "robot_banco_mundial.py"], capture_output=True, text=True, env=env)
    print(r.stdout + r.stderr)
    return r.returncode

def ver_bitacora(ultimas=12):
    archivo = Path(f"logs/bitacora_{EQUIPO}.jsonl")
    filas = [json.loads(l) for l in archivo.read_text(encoding="utf-8").splitlines() if l.strip()]
    t = pd.DataFrame(filas)
    cols = [c for c in ["momento_utc", "corrida", "nivel", "evento"] if c in t]
    display(t[cols].tail(ultimas))
    return t

escribir_robot(REGLAS)
codigo = correr_robot()
print("— Bitácora (las últimas líneas) —")
bit = ver_bitacora()
print("Cada línea tiene los mismos campos: cuándo, qué corrida, qué nivel y qué pasó.")
print("Niveles: INFO = normal · WARNING = algo raro pero siguió · ERROR = falló.")
print("Resultado:", "🟢 el robot terminó bien" if codigo == 0 else "🔴 el robot terminó con error (mira la última línea)")
print("✅ Paso 2 listo")
```

## PASO 3 · Las reglas de calidad del equipo

**👆 Qué haces:** 1. Lee la pista de abajo.
2. Ajusta los recuadros **MINIMO** y **MAXIMO** de cada indicador. Si no están seguros, dejen los valores que ya están.
3. En **MIN_FILAS** escribe cuántas filas, como mínimo, deben llegar (2 indicadores × 11 años = 22 como máximo).
4. En **MAX_ANIOS_SIN_DATO** decide cuántos años de antigüedad aceptan en el último dato.
5. Pulsa <img src="https://raw.githubusercontent.com/uide-programacion-inteligencia-mercados/recursos-curso/main/boton_ejecutar.png" width="22" alt="botón ejecutar" style="vertical-align:middle">. Puedes cambiar y volver a pulsar.

> **Ojo:** un rango muy estrecho da **falsas alarmas** (alertas por datos verdaderos). Un rango muy amplio deja pasar errores. La regla buena rechaza lo imposible, no lo raro.

## PASO 5 · Correr dos veces no daña nada (idempotencia)

**🔗 Lo que ya tienes:** el robot con sus reglas (Pasos 3 y 4).

**🎯 En este paso:** comprueban que su robot es **idempotente**: si corre dos veces, el resultado es el mismo. En GitHub esto pasa más de lo que creen: alguien pulsa *Run workflow* mientras el reloj ya lo había lanzado, o GitHub repite una corrida.

**👆 Qué haces:** pulsa el botón <img src="https://raw.githubusercontent.com/uide-programacion-inteligencia-mercados/recursos-curso/main/boton_ejecutar.png" width="22" alt="botón ejecutar" style="vertical-align:middle">. No escribes nada. Corre el robot dos veces como si fueran dos corridas distintas y luego **repite la misma corrida**.

**👀 Qué debes ver:** tres comprobaciones con ✅: el archivo de datos sale idéntico, la salud tiene una fila por corrida, y repetir la misma corrida no agrega una fila nueva. Al final: `✅ Paso 5 listo`.

**🧩 Lo que te llevas al paso siguiente:** la tranquilidad de que su robot no duplica ni ensucia datos.

```python
#@title PASO 5 · La prueba de las dos corridas { display-mode: "form" }
#@markdown Pulsa el botón redondo (no escribes nada)
import hashlib
if "escribir_robot" not in globals():
    print("✋ Todavía no puedes hacer este paso: primero pulsa el botón redondo del Paso 2.")
else:
    huella = lambda p: hashlib.md5(Path(p).read_bytes()).hexdigest()[:10] if Path(p).exists() else None
    datos, salud = Path(f"data/indicadores_{EQUIPO}.csv"), Path(f"logs/salud_{EQUIPO}.csv")
    print("Corrida A..."); a = correr_robot(GITHUB_RUN_ID="prueba-A")
    h1 = huella(datos)
    print("Corrida B..."); b = correr_robot(GITHUB_RUN_ID="prueba-B")
    h2 = huella(datos)
    filas_antes = len(pd.read_csv(salud))
    print("Repito la corrida B (como «Re-run jobs» en GitHub)..."); correr_robot(GITHUB_RUN_ID="prueba-B")
    filas_despues = len(pd.read_csv(salud))
    print("—" * 60)
    if a != 0 or b != 0:
        print("🔴 El robot terminó con error (¿sin internet?). Lean la última línea de arriba y vuelvan a pulsar en un minuto.")
    else:
        print(("✅" if h1 == h2 else "❌") + f" 1. Datos idénticos en las dos corridas (huella {h1} = {h2})")
        print("✅ 2. La salud guarda una fila por corrida: es un registro, no un dato que se reemplaza")
        print(("✅" if filas_antes == filas_despues else "❌") + f" 3. Repetir la MISMA corrida no agrega filas ({filas_antes} → {filas_despues})")
        display(pd.read_csv(salud).tail(4))
        print("Idempotente = correr una vez o varias deja el mismo resultado. Por eso nadie teme pulsar Run workflow.")
    print("✅ Paso 5 listo")
```

### Si algo sale mal
- **No aparece la casilla Simulacro:** la carpeta `.github` no se subió. Repite el punto A, paso 4.
- **La X roja aparece, pero no hay ninguna alerta en Issues:** abre la corrida roja y mira el paso *Abrir o actualizar la alerta*. Si dice «Resource not accessible» o algo parecido, falta un permiso: avisa al profesor por la Bandeja de Canvas con una captura.
- **No existe la pestaña Issues:** en tu casa, pulsa **Settings**, busca la sección **Features** y marca **Issues**.
- **La corrida 1 sale ROJA:** abre la fila, pulsa el trabajo `correr-robot` y busca la línea que empieza con `RESULTADO`. Si dice «datos no confiables», su regla del Paso 3 rechaza datos reales: amplíen el rango, vuelvan a armar el paquete y súbanlo de nuevo.
